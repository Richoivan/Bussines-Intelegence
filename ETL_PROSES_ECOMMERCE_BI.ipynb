```python
import pandas as pd

url = "https://github.com/Richoivan/Bussines-Intelegence/raw/main/E-Commerce-Train%20(1).csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#create column delay_Flag
df['Delay_Flag'] = df['Reached.on.Time_Y.N'].apply(lambda x: 'Late' if x == 1 else 'On Time')
```

```python
#create column weight Category
def weight_category(w):
    if w < 1500:
        return 'Light'
    elif w <= 3000:
        return 'Medium'
    else:
        return 'Heavy'

df['Weight_Category'] = df['Weight_in_gms'].apply(weight_category)
```

```python
#create column Discount Bucket
def discount_bucket(d):
    if d <= 20:
        return 'Low'
    elif d <= 50:
        return 'Medium'
    else:
        return 'High'

df['Discount_Bucket'] = df['Discount_offered'].apply(discount_bucket)
```

```python
#create column Rating Category
def rating_category(r):
    if r <= 2:
        return 'Low'
    elif r == 3:
        return 'Medium'
    else:
        return 'High'

df['Rating_Category'] = df['Customer_rating'].apply(rating_category)
```
